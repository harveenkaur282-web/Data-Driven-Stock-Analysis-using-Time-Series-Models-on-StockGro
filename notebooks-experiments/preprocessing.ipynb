# Task 2: Data Preprocessing
This notebook covers:
1. Loading raw historical data.
2. Stationarity testing (ADF Test).
3. Differencing for non-stationary series.
4. Feature scaling (MinMaxScaler) for Deep Learning models.
5. Train/Test splitting.

```python
import sys
sys.path.append("..")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from src.data.fetch_data import load_all_raw
from src.data.preprocess import preprocess_all
from statsmodels.tsa.stattools import adfuller

plt.style.use('seaborn-v0_8-whitegrid')
print("Environment Ready")
```

## 1. Load Data

```python
raw_data = load_all_raw()
ticker = "TCS.NS"
df = raw_data[ticker]
prices = df['Close']

plt.figure(figsize=(12, 5))
plt.plot(prices)
plt.title(f"{ticker} Raw Close Prices")
plt.show()
```

## 2. Stationarity Check (ADF Test)
If p-value > 0.05, the series is non-stationary and requires differencing.

```python
def check_stationarity(series):
    res = adfuller(series.dropna())
    print(f"ADF Statistic: {res[0]:.4f}")
    print(f"p-value: {res[1]:.4f}")
    return res[1] < 0.05

is_stat = check_stationarity(prices)
print(f"Is Stationary: {is_stat}")
```

## 3. Preprocessing Pipeline
Running the full pipeline defined in `src/data/preprocess.py`.

```python
processed = preprocess_all(raw_data, save=True)

tcs_data = processed[ticker]
print(f"\nProcessed data for {ticker}:")
print(f"Train size: {len(tcs_data['train'])}")
print(f"Test size: {len(tcs_data['test'])}")
print(f"Differencing order (d): {tcs_data['d_order']}")
```

## 4. Visualizing Scaled Data
Deep Learning models (LSTM/GRU) require data in [0, 1] range.

```python
plt.figure(figsize=(12, 4))
plt.plot(tcs_data['train_scaled'], label='Train Scaled', alpha=0.7)
plt.title(f"{ticker} Scaled Training Data")
plt.legend()
plt.show()
```
